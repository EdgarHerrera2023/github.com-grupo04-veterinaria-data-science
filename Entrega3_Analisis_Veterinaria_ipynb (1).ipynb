```python
import pandas as pd
import numpy as np

df = pd.read_csv("consultas_veterinaria.csv")
print(f"Se cargaron {len(df)} registros")
df.head()
```

```python
print("Cantidad de registros:", len(df))
print("\nColumnas:")
print(df.columns.tolist())
print("\nTipos de datos:")
print(df.dtypes)
```

```python
df[['costo', 'edad', 'peso']].describe()
```

```python
filtro_costo = df[df['costo'] > 15000]
print(f"Consultas con costo > $15.000: {len(filtro_costo)}")
filtro_costo[['fecha', 'nombre_mascota', 'motivo', 'costo']].head(10)
```

```python
filtro_perro = df[df['especie'] == 'Perro']
print(f"Consultas de perros: {len(filtro_perro)}")
filtro_perro[['fecha', 'nombre_mascota', 'raza', 'costo']].head(10)
```

```python
por_especie = df.groupby('especie').agg({
    'id_consulta': 'count',
    'costo': 'mean'
}).round(2)
por_especie.columns = ['cantidad_consultas', 'costo_promedio']
por_especie
```

```python
por_veterinario = df.groupby('veterinario')['id_consulta'].count().sort_values(ascending=False)
por_veterinario
```

```python
costos = df['costo'].to_numpy()

print("Operaciones con NumPy:")
print(f"Promedio:     ${np.mean(costos):.2f}")
print(f"Mediana:      ${np.median(costos):.2f}")
print(f"Desvío:       ${np.std(costos):.2f}")
print(f"Máximo:       ${np.max(costos):.2f}")
print(f"Mínimo:       ${np.min(costos):.2f}")
print(f"Suma total:   ${np.sum(costos):.2f}")
```

```python
print("="*60)
print("RESPUESTAS A LAS 5 PREGUNTAS DE ANÁLISIS")
print("="*60)

print("\n1. Promedio de costo de las consultas:")
print(f"   → ${df['costo'].mean():.2f}")

print("\n2. Consultas por especie:")
print(df.groupby('especie')['id_consulta'].count())

print("\n3. Veterinario con más consultas:")
print(df.groupby('veterinario')['id_consulta'].count().idxmax())

print("\n4. Costo máximo y mínimo:")
print(f"   Máximo: ${df['costo'].max():.2f}")
print(f"   Mínimo: ${df['costo'].min():.2f}")

print("\n5. Consultas por mes:")
print(df.groupby('mes')['id_consulta'].count().sort_index())
```

## Conclusiones

- El costo promedio de las consultas se encuentra en un rango intermedio.
- La mayor cantidad de atenciones corresponde a la especie **Perro**.
- Los veterinarios de Clínica General concentran la mayor cantidad de consultas.
- Existe una gran diferencia entre el costo mínimo (vacunas/controles) y el máximo (cirugías).
- El análisis permite a la clínica tomar mejores decisiones sobre recursos y demanda.
